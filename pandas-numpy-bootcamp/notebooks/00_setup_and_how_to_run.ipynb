## The sample data

The same fictional retail company as the other bootcamps, delivered as raw files.
If the next cell errors, run `uv run python data/build_data.py` in a terminal
first.

```python
# ▶ Run this first. Locates the sample data no matter where the kernel starts.
from pathlib import Path

def find_data() -> Path:
    here = Path.cwd()
    for base in (here, *here.parents):
        if (base / 'data' / 'raw').exists():
            return base / 'data'
    raise FileNotFoundError('Run: uv run python data/build_data.py')

DATA = find_data()
RAW = DATA / 'raw'
print('Data directory:', DATA)
print('Raw files:', sorted(p.name for p in RAW.glob('*')))
```

## A 30-second taste

NumPy and pandas operate on whole columns at once — no loops. Here is the
flavour; the notebooks explain every piece.

```python
import numpy as np, pandas as pd

prices = np.array([19.99, 5.0, 120.0])
print('with tax:', np.round(prices * 1.2, 2))     # vectorized, no loop

df = pd.read_csv(RAW / 'orders.csv')
print('orders:', len(df))
print(df.groupby('status')['amount'].sum().round(2))
```
