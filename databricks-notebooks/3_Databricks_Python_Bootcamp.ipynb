## 1 · Variables & types

A **variable** is a name bound to an object. Python is **dynamically typed**: you
never declare types, but every object has one (`type(x)` tells you). Assignment
never copies — names *refer* to objects.

````python
rows = 1500            # int
rate = 0.98            # float
name = "pipeline"      # str
ok   = True            # bool
nothing = None         # NoneType (Python's null)

print(rows, rate, name, ok, nothing)
print(type(rows), type(rate), type(name), type(ok))
````

````python
# Names are references. Mutable objects are shared; use .copy() for independence.
a = [1, 2, 3]
b = a                  # same list
b.append(4)
print("a:", a, "| a is b:", a is b)

c = a.copy()
c.append(99)
print("a:", a, "| c:", c)

# == compares VALUES, is compares IDENTITY. Use `is` only for None.
print([1,2] == [1,2], [1,2] is [1,2], nothing is None)
````

## 2 · Numbers

`int` has unlimited precision; `float` is a 64-bit approximation. Note the two
divisions: `/` (true, float), `//` (floor), `%` (remainder), `**` (power).
Floats are inexact — round money or use `decimal`.

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(2 ** 100)                      # big ints never overflow
print(0.1 + 0.2)                     # 0.30000000000000004  (float imprecision)
print(round(0.1 + 0.2, 2))

from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))   # exact 0.3
````

## 3 · Strings

Immutable text. **f-strings** embed expressions in `{}` with optional formatting
after `:`. Rich method set for cleaning; slice with `[start:stop:step]`.

````python
table, n, pct = "orders", 1523, 0.9834
print(f"Loaded {n:,} rows into {table} ({pct:.1%})")   # thousands + percent

s = "  Ava.Smith@Example.COM  "
print(s.strip().lower())
print("a,b,c".split(","), "-".join(["2024","01","15"]))
print("report.csv".endswith(".csv"), "Ava Smith".startswith("Ava"))

code = "ORD-2024-0042"
print(code[:3], code[-4:], code[::-1])       # slicing
prefix, year, seq = code.split("-")
print(prefix, int(year), int(seq))
````

## 4 · Booleans & truthiness

`if` tests *truthiness*, not just `True`/`False`. Empty containers, `0`, `""` and
`None` are **falsy**; everything else is **truthy** — handy for guarding missing
data.

````python
for v in [0, 1, "", "x", [], [1], None, {}]:
    print(f"{repr(v):>6} -> {bool(v)}")

email = ""
if not email:
    print("email is missing")
````

## 5 · Lists

Ordered, mutable sequences — the workhorse container. Grow, index, slice, sort.

````python
prices = [19.99, 5.0, 120.0, 5.0]
prices.append(42.0)
print(prices, "| len:", len(prices), "| max:", max(prices))
print("sorted:", sorted(prices))
prices.sort(reverse=True)
print("in place desc:", prices)
print("sum:", round(sum(prices), 2), "| first two:", prices[:2])
````

## 6 · Tuples

Ordered but **immutable**. Great for fixed records and multiple return values.
**Unpacking** assigns elements to names.

````python
row = (1042, "completed", 89.90)
order_id, status, amount = row      # unpacking
print(order_id, status, amount)

try:
    row[0] = 999
except TypeError as e:
    print("immutable:", e)
````

## 7 · Sets

Unordered collections of **unique** values — dedup and fast membership, plus set
algebra.

````python
countries = ["US", "GB", "US", "DE", "US"]
uniq = set(countries)
print(uniq, "| count:", len(uniq), "| 'US' in:", "US" in uniq)

a, b = {"US","GB","DE"}, {"DE","IN"}
print("intersection:", a & b, "| union:", a | b, "| only a:", a - b)
````

## 8 · Dictionaries

Key → value mappings — the shape of JSON and records. Access with `[]` (errors if
missing) or `.get()` (default). Iterate with `.items()`.

````python
cust = {"customer_id": 7, "name": "Ava Smith", "country": "US"}
print(cust["name"], "| email:", cust.get("email", "MISSING"))
cust["email"] = "ava@example.com"
for k, v in cust.items():
    print(" ", k, "=", v)

# Counting pattern + Counter
from collections import Counter
statuses = ["completed","returned","completed","cancelled","completed"]
print(Counter(statuses), "| top:", Counter(statuses).most_common(1))
````

## 9 · `namedtuple` & `deque`

Two specialized containers: a **`namedtuple`** is an immutable record with named
fields; a **`deque`** is a fast double-ended queue (great for sliding windows).

````python
from collections import namedtuple, deque

Order = namedtuple("Order", ["order_id", "amount", "status"])
o = Order(1042, 89.9, "completed")
print(o.order_id, o.amount, "|", o)

recent = deque(maxlen=3)             # keeps only the last 3
for i in range(6):
    recent.append(i)
print("sliding window:", list(recent))
````

## 10 · Control flow

Decisions with `if/elif/else`; repetition with `for` (over any iterable) and
`while`. `break` stops, `continue` skips. `enumerate` gives an index, `zip`
walks sequences in parallel, `range` generates numbers.

````python
amount = 250.0
tier = "enterprise" if amount >= 1000 else "standard" if amount >= 100 else "small"
print("tier:", tier)

total = 0
for p in [10, -5, 20, 0, 8]:
    if p <= 0:
        continue                      # skip invalid
    total += p
print("total of valid:", total)

for i, name in enumerate(["SSD","Book","Lego"], start=1):
    print(i, name)
for name, amt in zip(["Ava","Liam"], [89.9, 12.5]):
    print(name, amt)
````

## 11 · `match` — structural pattern matching (3.10+)

A clean way to branch on a value's shape — handy for routing events by type.

````python
def route(event):
    match event["type"]:
        case "page_view":
            return "analytics"
        case "checkout" | "add_to_cart":
            return "revenue"
        case _:
            return "other"

for t in ["page_view", "checkout", "search"]:
    print(t, "->", route({"type": t}))
````

## 12 · The walrus operator `:=`

Assign **and** return a value in one expression — avoids computing something
twice or repeating a read inside a loop/condition.

````python
data = [4, 9, 16, 25]
if (n := len(data)) > 3:
    print(f"{n} items - plenty")

queue = iter([5, 8, 0, 3])
while (v := next(queue, None)) is not None and v != 0:
    print("processing", v)
print("stopped")
````

## 13 · Comprehensions

Build a collection from an iterable in one expression — the idiomatic replacement
for "init empty, loop, append". Add `if` to filter; use `()` for a lazy
generator.

````python
prices = [10, 25, 5, 100, 3]
print([round(p * 1.2, 2) for p in prices])          # list comp
print([p for p in prices if p >= 25])                # filter
print({p: p*p for p in [2,3,4]})                     # dict comp
print({c[0] for c in ["Books","Home","Toys"]})       # set comp

total = sum(p for p in prices if p >= 25)            # generator expr (lazy)
print("sum>=25:", total)
````

## 14 · Functions

Package logic behind a name. Parameters can have **defaults**; `*args`/`**kwargs`
collect extra positional/keyword arguments; **type hints** document intent (not
enforced). ⚠️ Never use a *mutable* default — use `None`.

````python
def net_amount(gross: float, tax_rate: float = 0.2) -> float:
    "Return gross plus tax, rounded to cents."
    return round(gross * (1 + tax_rate), 2)

print(net_amount(100), net_amount(100, 0.05), net_amount(gross=100, tax_rate=0.1))

def summarize(*values, **options):
    label = options.get("label", "total")
    return f"{label}: {sum(values)}"
print(summarize(10, 20, 30, label="revenue"))
print(summarize(*[1,2,3,4]))               # unpack a list into args

# Safe mutable default via None sentinel
def add_item(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket
print(add_item("a"), add_item("b"))        # ['a'] then ['b'] (independent)
````

## 15 · Scope, closures & lambda

Inner functions can **capture** variables from the enclosing scope (a
**closure**). `lambda` is a one-expression anonymous function, handy as a sort
`key`.

````python
def make_multiplier(factor):
    def multiply(x):
        return x * factor          # captures `factor`
    return multiply
double, triple = make_multiplier(2), make_multiplier(3)
print(double(10), triple(10))

orders = [{"id":1,"amount":89.9},{"id":2,"amount":240.0},{"id":3,"amount":12.5}]
print([o["id"] for o in sorted(orders, key=lambda o: o["amount"], reverse=True)])
````

## 16 · Decorators

A **decorator** wraps a function to add behavior without changing its code
(built on closures). `@name` means `func = name(func)`. Use `functools.wraps` to
keep the wrapped function's identity. Ready-made: `functools.lru_cache`.

````python
import functools, time

def timed(fn):
    @functools.wraps(fn)
    def wrapper(*args, **kwargs):
        t0 = time.perf_counter()
        try:
            return fn(*args, **kwargs)
        finally:
            print(f"{fn.__name__} took {(time.perf_counter()-t0)*1000:.1f} ms")
    return wrapper

@timed
def crunch(n):
    return sum(i*i for i in range(n))
print("result:", crunch(200_000))

def retry(times=3):
    def deco(fn):
        @functools.wraps(fn)
        def wrapper(*a, **k):
            for attempt in range(1, times+1):
                try:
                    return fn(*a, **k)
                except Exception as e:
                    print(f"attempt {attempt} failed: {e}")
                    if attempt == times: raise
        return wrapper
    return deco

_n = {"c": 0}
@retry(times=3)
def flaky():
    _n["c"] += 1
    if _n["c"] < 3: raise ConnectionError("transient")
    return "ok"
print("final:", flaky())

@functools.lru_cache(maxsize=None)
def slow_square(n):
    print(f"  computing {n}")
    return n*n
print(slow_square(12), slow_square(12))    # second call is cached
````

## 17 · Iterators & generators

Process data lazily, one item at a time — the key to handling data bigger than
memory. A function with `yield` is a **generator**; `itertools` adds fast
building blocks.

````python
def countdown(n):
    while n > 0:
        yield n
        n -= 1
print(list(countdown(5)))

# Lazy pipeline: filter -> transform -> aggregate, constant memory
records = ({"id": i, "amount": (i*7) % 100} for i in range(1_000_000))
big = (r for r in records if r["amount"] >= 50)
total = sum(r["amount"] for r in big)
print("total:", total)

import itertools
print("islice:", list(itertools.islice(range(1_000_000), 5)))
print("chain:", list(itertools.chain([1,2],[3,4])))
rows = [("US",1),("US",2),("GB",3),("GB",4)]
for k, g in itertools.groupby(rows, key=lambda r: r[0]):
    print(k, [x[1] for x in g])
````

## 18 · Modules & the standard library

`import` brings in modules; Python ships with a huge **standard library** so much
is already there.

````python
import math, statistics as stats, random, uuid
from collections import defaultdict

print(math.ceil(4.2), math.sqrt(144), round(math.pi, 4))
print("mean:", stats.mean([10,20,30]), "| median:", stats.median([1,2,3,100]))

random.seed(0)
print("random pick:", random.choice(["a","b","c"]))
print("a unique id:", uuid.uuid4())

groups = defaultdict(list)             # never worry about missing keys
for country, cid in [("US",1),("GB",2),("US",3)]:
    groups[country].append(cid)
print("grouped:", dict(groups))
````

## 19 · Object-Oriented Programming

A **class** bundles data (attributes) with behavior (methods). `__init__` is the
constructor; **dunder** methods hook into Python syntax (`__repr__`, `__eq__`).

````python
class Accumulator:
    "Running total and count."
    def __init__(self):
        self.total, self.count = 0.0, 0
    def add(self, amount):
        self.total += amount
        self.count += 1
    def average(self):
        return self.total / self.count if self.count else 0.0

acc = Accumulator()
for a in [10, 20, 30]:
    acc.add(a)
print("total:", acc.total, "| avg:", acc.average())

class Money:
    def __init__(self, cents): self.cents = cents
    def __repr__(self): return f"Money(${self.cents/100:.2f})"
    def __eq__(self, other): return self.cents == other.cents
    def __add__(self, other): return Money(self.cents + other.cents)

print(Money(1099) + Money(500), "| equal:", Money(500) == Money(500))
````

## 20 · Dataclasses, inheritance & properties

`@dataclass` generates `__init__`/`__repr__`/`__eq__` for record types.
**Inheritance** shares an interface; `@property` exposes a computed value like an
attribute.

````python
from dataclasses import dataclass, field

@dataclass
class Customer:
    customer_id: int
    name: str
    country: str = "US"
    tags: list = field(default_factory=list)   # safe mutable default

c1, c2 = Customer(1,"Ava"), Customer(1,"Ava")
print(c1, "| equal:", c1 == c2)

class Loader:
    def load(self, rows): raise NotImplementedError
class ConsoleLoader(Loader):
    def load(self, rows):
        for r in rows: print("LOAD", r)
ConsoleLoader().load([{"id":1},{"id":2}])
print("is a Loader:", isinstance(ConsoleLoader(), Loader))

class Order:
    def __init__(self, qty, price): self.qty, self.price = qty, price
    @property
    def total(self): return round(self.qty * self.price, 2)
print("total:", Order(3, 19.99).total)          # no parentheses
````

## 21 · Errors & exceptions

When something goes wrong Python **raises** an exception. Catch specific ones with
`try/except`; `else` runs on success, `finally` always runs. Raise your own for
clear failures.

````python
def to_int(v):
    try:
        return int(v)
    except ValueError:
        return None            # couldn't parse -> treat as missing
print([to_int(x) for x in ["42", "3.5", "", "abc", " 7 "]])

class DataQualityError(Exception):
    "Raised when a record fails validation."

def validate(order):
    if order["amount"] < 0:
        raise DataQualityError(f"negative amount in order {order['id']}")
    return order

try:
    validate({"id": 5, "amount": -10})
except DataQualityError as e:
    print("rejected:", e)
finally:
    print("...cleanup always runs")
````

## 22 · Context managers (`with`)

A context manager guarantees setup/teardown even if an error occurs — the classic
case is closing files. Build your own with `contextlib.contextmanager`.

````python
import time
from contextlib import contextmanager

@contextmanager
def timer(label):
    start = time.perf_counter()
    try:
        yield                                   # <- the `with` body runs here
    finally:
        print(f"{label}: {(time.perf_counter()-start)*1000:.1f} ms")

with timer("sum 1M"):
    s = sum(range(1_000_000))
print("result:", s)
````

## 23 · Files & `pathlib`

Read and write files with `pathlib.Path`. We use a temp directory so this runs
anywhere with nothing to clean up. Always pass `encoding="utf-8"`.

````python
import tempfile
from pathlib import Path

path = Path(tempfile.gettempdir()) / "demo.txt"
path.write_text("line 1\nline 2\nline 3\n", encoding="utf-8")

text = path.read_text(encoding="utf-8")
print("lines:", len(text.splitlines()))
for line in text.splitlines():
    print(" ", line)
print("name:", path.name, "| suffix:", path.suffix, "| exists:", path.exists())
````

## 24 · JSON

APIs, config and logs are mostly JSON. `loads`/`dumps` convert between JSON text
and Python objects (dict/list/str/num/bool/None).

````python
import json

text = '{"id": 7, "tags": ["vip","eu"], "active": true, "note": null}'
obj = json.loads(text)                 # JSON -> dict
print(type(obj), obj["tags"], obj["active"], obj["note"])

back = json.dumps({"rows": 1523, "ok": True}, indent=2)   # dict -> JSON
print(back)

# JSON Lines (one object per line) - the event-stream format
lines = '\n'.join(json.dumps({"i": i}) for i in range(3))
print([json.loads(l) for l in lines.splitlines()])
````

## 25 · CSV

The `csv` module parses delimited text correctly (respecting quotes). We use an
in-memory buffer so there's no file to upload.

````python
import csv, io

raw = "order_id,status,amount\n1001,completed,89.90\n1002,returned,14.00\n"
reader = csv.DictReader(io.StringIO(raw))
rows = list(reader)
print("columns:", reader.fieldnames)
revenue = sum(float(r["amount"]) for r in rows if r["status"] == "completed")
print("completed revenue:", revenue)

out = io.StringIO()
w = csv.DictWriter(out, fieldnames=["status","n"])
w.writeheader()
w.writerow({"status": "completed", "n": 1})
print(out.getvalue())
````

## 26 · Dates & times

`datetime`/`date`/`timedelta` model instants, days and durations. `strptime`
parses strings, `strftime` formats them (and builds partition paths).

````python
from datetime import date, datetime, timedelta

today = date(2024, 6, 15)
print(today, today.strftime("%A"), "| +30d:", today + timedelta(days=30))

dt = datetime.strptime("2024-06-01 14:30:00", "%Y-%m-%d %H:%M:%S")
print(dt.year, dt.hour)
print(dt.strftime("year=%Y/month=%m/day=%d"))     # Hive-style partition path

start = datetime(2024,6,1,9,0); end = datetime(2024,6,1,17,30)
print("hours:", (end - start).total_seconds() / 3600)
````

## 27 · Regular expressions

`re` matches patterns in text — extract and validate fields that aren't cleanly
delimited. Use **raw strings** (`r"..."`).

````python
import re

text = "Order ORD-2024-0042 shipped"
m = re.search(r"ORD-(\d{4})-(\d+)", text)
print("match:", m.group(), "| year:", m.group(1), "| seq:", int(m.group(2)))

print(re.findall(r"\d+", "a1 b22 c333"))                       # ['1','22','333']
print(re.sub(r"[^0-9]", "", "ID: 00-1234"))                     # keep digits
email_re = re.compile(r"^[\w.+-]+@[\w-]+\.[\w.-]+$")
for a in ["ava.smith@example.com", "not-an-email"]:
    print(f"{a:25s} valid={bool(email_re.fullmatch(a))}")
````
