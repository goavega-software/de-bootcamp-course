# 03 · Sorting & Limiting

- `ORDER BY` sorts results (`ASC` ascending default, `DESC` descending)
- sort by multiple columns
- `LIMIT` + `OFFSET` for "top N" and pagination

```python
# ▶ Run this cell first. It loads JupySQL and connects to the SQLite database.
%load_ext sql
from sqlalchemy import create_engine
import os

# Works whether the notebook's working dir is the repo root or notebooks/
db_path = 'data/retail.db' if os.path.exists('data/retail.db') else '../data/retail.db'
engine = create_engine(f'sqlite:///{db_path}')

%config SqlMagic.autopandas = True      # results come back as pandas DataFrames
%config SqlMagic.displaycon = False
%config SqlMagic.feedback = 0
%config SqlMagic.displaylimit = 100

%sql engine
print('Connected to', db_path)
```

## Sort ascending (default)

```python
%%sql
SELECT product_name, unit_price FROM products ORDER BY unit_price;
```

## Sort descending — most expensive first

```python
%%sql
SELECT product_name, unit_price FROM products ORDER BY unit_price DESC;
```

## Sort by multiple columns
By category, then by price within each category:

```python
%%sql
SELECT category_id, product_name, unit_price
FROM products
ORDER BY category_id ASC, unit_price DESC;
```

## Top-N with `LIMIT`
The 3 most expensive products:

```python
%%sql
SELECT product_name, unit_price
FROM products
ORDER BY unit_price DESC
LIMIT 3;
```

## Pagination with `OFFSET`
Skip the first 3, then take the next 3 (page 2):

```python
%%sql
SELECT product_name, unit_price
FROM products
ORDER BY unit_price DESC
LIMIT 3 OFFSET 3;
```
