# 07 · Set Operations

Stack the results of two queries vertically. Both queries must return the same
number of columns with compatible types.
- `UNION` — combine and remove duplicates
- `UNION ALL` — combine and keep duplicates (faster)
- `INTERSECT` — rows in both
- `EXCEPT` — rows in the first but not the second

```python
# ▶ Run this cell first. It loads JupySQL and connects to the SQLite database.
%load_ext sql
from sqlalchemy import create_engine
import os

# Works whether the notebook's working dir is the repo root or notebooks/
db_path = 'data/retail.db' if os.path.exists('data/retail.db') else '../data/retail.db'
engine = create_engine(f'sqlite:///{db_path}')

%config SqlMagic.autopandas = True      # results come back as pandas DataFrames
%config SqlMagic.displaycon = False
%config SqlMagic.feedback = 0
%config SqlMagic.displaylimit = 100

%sql engine
print('Connected to', db_path)
```

## `UNION ALL`
A combined contact list of customer and employee names, tagged by source:

```python
%%sql
SELECT first_name, last_name, 'customer' AS kind FROM customers
UNION ALL
SELECT first_name, last_name, 'employee' AS kind FROM employees
ORDER BY kind, last_name;
```

## `UNION` (dedupes)
All distinct countries where we have either a customer or a supplier:

```python
%%sql
SELECT country FROM customers
UNION
SELECT country FROM suppliers
ORDER BY country;
```

## `INTERSECT`
Countries that have **both** a customer and a supplier:

```python
%%sql
SELECT country FROM customers
INTERSECT
SELECT country FROM suppliers;
```

## `EXCEPT`
Countries with a customer but **no** supplier:

```python
%%sql
SELECT country FROM customers
EXCEPT
SELECT country FROM suppliers;
```
