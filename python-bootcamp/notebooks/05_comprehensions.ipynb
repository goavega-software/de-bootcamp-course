# 05 · Comprehensions

A **comprehension** builds a collection from an iterable in one readable
expression. They replace the common "make an empty list, loop, append" pattern
and are the idiomatic way to transform and filter data in Python.

## List comprehensions

Read it as: *expression* **for** *item* **in** *iterable* (**if** *condition*).

```python
prices = [10, 25, 5, 100, 3]

# the long way
with_tax = []
for p in prices:
    with_tax.append(round(p * 1.2, 2))
print('loop:', with_tax)

# the comprehension way
with_tax = [round(p * 1.2, 2) for p in prices]
print('comp:', with_tax)
```

## Filtering with `if`

Add a trailing `if` to keep only items that match — cleaning and selecting rows
in one line.

```python
amounts = [120, -5, 89, 0, 240, -12]
valid = [a for a in amounts if a > 0]
print('valid:', valid)

# conditional expression INSIDE (transform every item, two ways)
labels = ['big' if a >= 100 else 'small' for a in valid]
print('labels:', labels)
```

## Dict and set comprehensions

Same syntax, different braces. Build lookup tables and unique sets directly.

```python
products = [('ELE-001', 19.99), ('BOO-014', 8.5), ('GRO-003', 3.2)]

price_by_sku = {sku: price for sku, price in products}
print(price_by_sku)
print('lookup BOO-014:', price_by_sku['BOO-014'])

categories = {'Electronics', 'Books', 'Books', 'Home'}
initials = {c[0] for c in categories}     # set comprehension
print('initials:', initials)
```

## Generator expressions

Swap `[]` for `()` and you get a **generator**: it produces items lazily, one at
a time, without building the whole list in memory. Perfect for aggregating over
huge inputs (you'll go deeper in the generators notebook).

```python
amounts = [120, 89, 240, 15, 60]

total = sum(a for a in amounts if a >= 50)   # no intermediate list
print('total >= 50:', total)

gen = (a * 2 for a in amounts)
print(type(gen))
print('first value:', next(gen))            # pulled on demand
```

## Nested comprehensions (flatten)

Two `for` clauses flatten nested structure. Keep it to one or two levels — past
that, a normal loop reads better.

```python
orders = [['ELE-001', 'BOO-014'], ['GRO-003'], ['TOY-009', 'HOM-002']]
all_skus = [sku for order in orders for sku in order]
print(all_skus)
```

### Recap

Comprehensions turn "init, loop, append" into one expression; add `if` to
filter; `{}` builds dicts/sets; `()` builds a lazy generator for big data.
